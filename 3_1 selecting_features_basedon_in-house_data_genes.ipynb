```python
import pandas as pd
```

```python
house_data = pd.read_csv("data/merged_original_data.csv")
print("inhouse_data.shape: ", house_data.shape)
genes_list = list(house_data['Gene'])

processed_data = pd.read_csv("data/ready2analyze.csv")
print("processed_data.shape: ", processed_data.shape)
```

```python
processed_data.head(5)
```

```python

processed_genes = processed_data.columns.tolist()
processed_genes

processed_genes.remove('outcomes')

processed_genes

to_remove = []
for i in range(len(processed_genes)):
    if processed_genes[i] in genes_list:
        #print("gene existsed: ", processed_genes[i])
        pass
    else:
        print("gene not existed: ", processed_genes[i])
        to_remove.append(processed_genes[i])
        
processed_data.head(5)
```

```python
in_house_temp= processed_data.drop(to_remove, axis=1)
```

```python
in_house = in_house_temp
in_house.head()
```

```python
in_house.to_csv("data/processed_inhouse.csv", index=False)
```
