```python
import pandas as pd
```

```python
mrg_data = pd.read_csv("merged.csv", header=None, index_col=False).T
```

```python
print("mrg_data.shape: ", mrg_data.shape)
mrg_data.head(5)
```

```python
df = mrg_data
#set column names equal to values in row index position 0
df.columns = df.iloc[0]
#remove first row from DataFrame
df = df[1:]
print("df.shape: ", df.shape)
df.head(5)
```

```python
nan_count = df.isna().sum()
threshold = 2
columns_to_drop = nan_count[nan_count > threshold].index
df_new = df.drop(columns=columns_to_drop)
```

```python
nan_count = df_new.isna().sum()
```

```python
df_new.shape
```

```python
df_new2 = df_new.dropna()
```

```python
nan_count = df_new2.isna().sum()
```

```python
df_new3 = df_new2.drop(["Gene"], axis=1)
```

```python
df_new3.head(5)
```

```python
df_new3.to_csv("processed_data/merged_formatted.csv", index=False)
```

```python
selected_features = ['ZFP36L2', 'CX3CR1', 'SLC12A7', 'GIMAP4', 'A2M', 'FMN1', 'IQGAP1', 'MTRNR2L10']
new_mrg_data = df_new3[selected_features]
```

```python
new_mrg_data.to_csv("new_mrg_data.csv", index=False)
```
