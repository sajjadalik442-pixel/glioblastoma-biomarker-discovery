```python
col_names = list(data.columns)
col_names
col_names.pop(-1)
```

```python
c = 1 #number of unique values
count = 0 #count of columns having "c" number of unique values - in each column 
u_labels = []
for col_name in col_names:
    if data[col_name].nunique() ==c:
        count = count +1
        u_labels.append(col_name)
        #print(col_name)
    else:
        pass
    
print(count)
```

```python
print("data.shape: ", data.shape)
data_drop = data.drop(u_labels, inplace=False, axis = 1)
print("data_drop.shape: ", data_drop.shape)
```

```python
data_drop.to_csv("data/cardinality.csv", index=False)
```
